### 1. Algoritmos de Ordenación y Búsqueda

En el mundo de la programación existen algoritmos de ordenación y de búsqueda, los cuales se encargan de ordenar estructuras de datos y de hacer búsquedas dentro de las mismas.

#### Algoritmo Burbuja o Bubble Sort

Es un algoritmo que se encarga de ir comparando los números de 2 en 2 y revisar cuál de ellos es el mayor, y así intercambiar de posición en caso de que el primer número sea mayor al segundo. Así, el mayor número pasará a estar a la derecha de la lista o, si lo vemos de forma vertical, pasará a estar arriba del todo. Viendo esta analogía, es como si las burbujas (número mayor) de un refresco flotaran hacia la superficie; por esto se le llama algoritmo burbuja. 



Veamos un ejemplo aplicado a Python.

**Desglose lógico del código:**
* `cambio = True`: Declaramos una variable con el valor `True` para poder entrar al bucle inicial.
* `longitud = len(lista)`: Nos ayuda a delimitar el rango de la lista sin calcularlo en cada iteración.
* `while cambio:` Como la variable es `True`, iniciamos el ciclo.
* `cambio = False`: Inmediatamente asumimos que no hay cambios. Esta misma asignación es la que nos sacará del ciclo `while` cuando nuestra lista esté completamente ordenada, rompiendo el bucle.
* `for i in range(longitud - 1):` Iniciamos un bucle `for` para iterar desde el primer elemento hasta el penúltimo. Esto se hace porque si iteráramos hasta el último número, al sumarle 1 saldríamos del rango de la lista (ocasionando un *IndexError*).
* `if lista[i] > lista[i + 1]:` Comparamos el número de la izquierda con el de la derecha. El `+ 1` asegura que el penúltimo se compare con el último.
* `lista[i], lista[i + 1] = lista[i + 1], lista[i]`: Si la condición se cumple, aplicamos asignación múltiple para intercambiarlos de forma fácil.
* `cambio = True`: Si hubo un intercambio, devolvemos el valor a `True` para obligar al ciclo `while` a dar otra vuelta revisando toda la lista nuevamente.

```python
lista = [58, 73, 24, 87, 12, 39, 65, 5, 18, 92, 42, 88, 33, 1, 79, 45, 81, 70, 23, 60]
cambio = True
longitud = len(lista)

while cambio:
    cambio = False 
    for i in range(longitud - 1): 
        if lista[i] > lista[i + 1]: 
            lista[i], lista[i + 1] = lista[i + 1], lista[i] 
            cambio = True 

print("Esta es la lista ordenada con Bubble sort: ", lista)
```

> **Anotación del Profesor:**
> Has documentado la lógica matemática del Algoritmo Burbuja a la perfección. Comprender esto te da bases sólidas en estructuras de datos, pero como futuro desarrollador Back End debes saber que este algoritmo es sumamente ineficiente computacionalmente para grandes volúmenes de datos. Para solucionar esto, los lenguajes modernos utilizan otros métodos nativos por debajo.

### 2. Métodos Nativos de Python

Ten en mente esto: siempre es importante aprender las bases para saber cómo funcionan las cosas de manera simplificada. Ahora que ya sabes cómo funciona un algoritmo de ordenamiento simple, podemos utilizar las herramientas que Python nos brinda, y para esto tenemos los métodos `.sort()` y `.reverse()` para ordenar nuestra lista y para invertirla. Veamos cómo funciona.

> **Anotación del Profesor:**
> Python utiliza internamente un algoritmo llamado **Timsort** (una implementación híbrida altamente optimizada en C) cuando llamas a `.sort()`. Recuerda una regla de oro: `.sort()` modifica la lista original *in-place* (ahorrando memoria RAM), no devuelve una lista nueva.

```python
lista = [58, 73, 24, 87, 12, 39, 65, 5, 18, 92, 42, 88, 33, 1, 79, 45, 81, 70, 23, 60]

lista.sort()
print("Lista ordenada con .sort(): ", lista)

lista.reverse()
print("Lista invertida con .reverse(): ", lista)
```

**Nota:** Los métodos anteriores también funcionan perfectamente para listas de *strings* (cadenas de texto), ordenándolas de manera alfabética.

```python
nombres_mascotas = [
    "Luna", "Max", "Bella", "Rocky", "Lucy",
    "Charlie", "Milo", "Daisy", "Toby", "Nala",
    "Simba", "Cooper", "Mia", "Buddy", "Chloe"
]

nombres_mascotas.sort()
print("Lista de mascotas ordenada: ", nombres_mascotas)

nombres_mascotas.reverse()
print("Lista de mascotas invertida: ", nombres_mascotas)
```
