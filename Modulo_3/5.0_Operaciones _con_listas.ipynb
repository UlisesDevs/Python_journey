### 1. Copiando Listas: Referencias vs. Copias Reales

¿Cuando tratamos de copiar una lista podemos asignar una nueva variable con la variable que contiene nuestra lista original? Vamos a ver si es posible.

```python
list_1 = [1]
list_2 = list_1
list_1[0] = 2

print(list_2)
```

Te darás cuenta de que el resultado al momento de imprimir nuestra lista es `2` y no `1`. 

Esto es porque al copiarla de esa manera, copiamos el espacio en memoria asignado a esa lista; por lo tanto, cualquier cambio que realicemos en nuestra primera lista se reflejará en la segunda. ¿Pero qué pasa si lo que nosotros queremos hacer es copiar la primera lista sin que en la segunda se realicen los cambios que se le hagan a la primera? Para eso tenemos el operador slice `:`.

> **Anotación del Profesor:**
> En Python, las variables no son "cajas" que guardan datos, sino "etiquetas" (punteros o referencias) que apuntan a un objeto en la memoria. Al hacer `list_2 = list_1`, no estás creando una lista nueva, solo estás pegando una segunda etiqueta al mismo objeto. Esto se conoce como **Aliasing**. 

### 2. Operador Slice `[:]`

Este operador nos sirve para copiar el contenido de una lista a una nueva lista separada en memoria.

```python
list_1 = [1]
list_2 = list_1[:]
list_1[0] = 2

print("Imprime la lista 2 copiada con slice:", list_2)

lista = [2, 6, 8, 34, 0, 12]
lista_2 = lista[:]
lista.append(54)

print("Imprime la lista 1 (original modificada):", lista)
print("Imprime la lista 2 (copia intacta):", lista_2)
```

Ahora verás que el resultado al imprimir nuestra segunda lista esta vez es `1` y no `2` como anteriormente. Utilizar el operador de esta forma nos servirá para copiar la lista entera.

> **Anotación del Profesor:**
> Utilizar `[:]` crea lo que se conoce como una **Shallow Copy** (copia superficial). Crea un nuevo objeto de lista en memoria, pero si tuvieras listas anidadas (una lista dentro de otra lista), los elementos internos seguirían compartiendo la misma referencia.

### 3. Copiando Porciones de una Lista

También podemos copiar porciones de la lista con la siguiente sintaxis: `lista_a_copiar[primer_indice:ultimo_indice]`

**Nota:** No se copiará el último índice, sino uno anterior al especificado.

```python
lista_3 = lista[0:4]
print("Imprime la lista 3:", lista_3)

lista_4 = lista[2:-1]
print("Imprime la lista 4 (con índices negativos):", lista_4)
```

> **Anotación del Profesor:**
> La regla estricta en Python para el particionado (slicing) es que el límite inferior es **inclusivo** y el límite superior es **exclusivo**. Por eso el último índice declarado no se incluye en el resultado.

Si se omite el primer índice, se infiere que queremos copiar desde el índice `0`. Sería lo mismo que escribir `[0:4]`.

```python
lista_5 = lista[:3]
print("Imprime la lista 5 (omitiendo inicio):", lista_5)
```

Por otro lado, si lo que se omite es el último índice, se infiere que queremos copiar hasta el final de la lista. Sería lo mismo que escribir `[1:len(lista_original)]`.

```python
lista_6 = lista[2:]
print("Imprime la lista 6 (omitiendo final):", lista_6)
```

### 4. Borrando Contenido con Slices y `del`

Para borrar contenido de una lista con el operador slice podemos utilizar la palabra reservada `del` con la misma sintaxis: `del mi_lista[1:3]`.

> **Anotación del Profesor:**
> Aunque `del lista[:]` funciona perfectamente para vaciar una lista, en el desarrollo profesional la forma más moderna, legible y recomendada (introducida en versiones más recientes de Python) es utilizar el método nativo: `lista.clear()`. Sin embargo, `del` sigue siendo la herramienta correcta para borrar rangos específicos.

```python
lista_para_borrar = [2, 6, 8, 34, 0, 12]
print("Lista original:", lista_para_borrar)

del lista_para_borrar[:]
print("Lista borrada por completo:", lista_para_borrar)

lista_para_borrar_rangos = [2, 6, 8, 34, 0, 12]
del lista_para_borrar_rangos[1:4]
print("Lista borrada del índice 1 al 4:", lista_para_borrar_rangos)
```

### 5. Comprobando la Existencia de Elementos

Para comprobar si hay o no elementos dentro de una lista utilizaremos las palabras reservadas `in` y `not in`, las cuales nos regresarán un Booleano (`True` o `False`). Su sintaxis es:

* `elemento in my_list`
* `elemento not in my_list`

> **Anotación del Profesor:**
> Este conocimiento es crucial para el rendimiento en el Back End. Buscar un elemento en una lista con `in` obliga a Python a revisar elemento por elemento de izquierda a derecha. Esto tiene una complejidad de tiempo lineal, denotada matemáticamente como $O(n)$. Si necesitas realizar cientos de búsquedas en una base de datos o estructura de millones de registros, es mejor convertir la lista a un `set` (conjunto). Las búsquedas en los *sets* se realizan mediante tablas hash, lo que reduce la complejidad a un tiempo constante de $O(1)$, siendo una operación prácticamente instantánea sin importar el tamaño de los datos.

```python
my_list = [0, 3, 12, 8, 2]
print("Lista de comprobación:", my_list)

print("¿5 está en la lista?:", 5 in my_list)
print("¿5 no está en la lista?:", 5 not in my_list)
print("¿12 está en la lista?:", 12 in my_list)
```
