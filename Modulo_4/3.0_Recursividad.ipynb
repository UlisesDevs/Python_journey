### 1. Funciones Recursivas

Las funciones recursivas son aquellas que se llaman a sí mismas como parte de su proceso de ejecución. Esto se utiliza para resolver problemas que pueden ser divididos en subproblemas más pequeños del mismo tipo. En esencia, una función recursiva resuelve un problema llamándose a sí misma hasta que se alcanza una **condición base** (o caso base) que detiene la recursión.

#### Características de una Función Recursiva:

1. **Caso base (condición de parada):** Es la condición que indica cuándo la función debe dejar de llamarse a sí misma. Sin un caso base, la recursión continuará indefinidamente, provocando un desbordamiento de la pila de llamadas en memoria (*Stack Overflow*).
2. **Caso recursivo:** Es el escenario donde la función sigue llamándose a sí misma, pero con una versión más simple o reducida del problema inicial.



### 2. Ejemplo Básico: Factorial de un Número

El factorial de un número $n$ se define como el producto de todos los números enteros positivos hasta $n$. Por ejemplo, el factorial de 5 (escrito como $5!$) es:

$$5! = 5 \times 4 \times 3 \times 2 \times 1 = 120$$

Esta es una definición naturalmente recursiva, ya que podemos abstraer la fórmula general de la siguiente manera:

$$n! = n \times (n - 1)!$$

```python

def factorial(n):
    # Caso base: Si n es 0 o 1, el factorial es 1 (matemáticamente 0! = 1)
    if n == 0 or n == 1:
        return 1
    # Caso recursivo: n multiplicado por el factorial de n-1
    else:
        return n * factorial(n - 1)

print("El factorial de 5 es:", factorial(5))
```

### 3. Ejemplo: Serie de Fibonacci

La serie de Fibonacci es una secuencia de números donde cada número es la suma de los dos anteriores. La secuencia empieza con 0 y 1, y se define matemáticamente como:

* $F(0) = 0$
* $F(1) = 1$
* $F(n) = F(n - 1) + F(n - 2)$ para $n \ge 2$

```python

def fibonacci(n):
    # Caso base: F(0) = 0, F(1) = 1
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Caso recursivo: Suma de los dos números anteriores en la serie
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print("La posición 6 de la serie Fibonacci es:", fibonacci(6)) 
# Secuencia: 0, 1, 1, 2, 3, 5, 8
```

### 4. Consideraciones sobre Recursión

* **Complejidad y Rendimiento:** La recursión es a menudo más intuitiva para problemas divisibles, pero puede ser menos eficiente en términos de memoria y tiempo de ejecución. Por ejemplo, el cálculo de Fibonacci con recursión pura es extremadamente ineficiente porque recalcula los mismos valores una y otra vez formando un árbol de llamadas exponencial.



* **Optimización (Memoization):** Es una técnica para optimizar funciones almacenando en caché (por ejemplo, en un diccionario) los resultados de subproblemas ya resueltos, evitando procesarlos de nuevo.
* **Recursión vs. Iteración:** Muchas funciones recursivas se pueden reescribir utilizando un bucle iterativo (ciclos `for` o `while`). La recursión es más limpia de leer, pero la iteración suele ser más eficiente en consumo de RAM porque no satura la pila de llamadas del sistema operativo.

> **Anotación del Profesor:**
> En tus apuntes originales definiste la función así: `def fibonacci_memo(n, memo={}):`. Esto es un **anti-patrón gravísimo en Python** conocido como *Mutable Default Argument*. 
> 
> En Python, los argumentos por defecto (como ese diccionario vacío) se evalúan y crean **una sola vez** cuando se define la función, no cada vez que la llamas. Esto significa que si llamas a la función varias veces a lo largo de tu programa para distintos propósitos, todas compartirán el mismo diccionario `memo` en memoria, provocando fugas de datos y resultados erróneos. La forma correcta de manejar estructuras mutables como parámetros por defecto es usar `None` e inicializarlas dentro del cuerpo de la función.

```python
# FORMA CORRECTA Y SEGURA DE DECLARAR EL CACHÉ:
def fibonacci_memo(n, memo=None):
    # Si memo no existe en la llamada inicial, lo creamos de forma local
    if memo is None:
        memo = {}
        
    # Casos base
    if n == 0:
        return 0
    elif n == 1:
        return 1
        
    # Si el resultado ya fue calculado previamente y está en memoria, lo retornamos directo
    if n in memo:
        return memo[n]
        
    # Caso recursivo: calculamos y almacenamos en la caché local al mismo tiempo
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    
    return memo[n]

print("Fibonacci optimizado (posición 6):", fibonacci_memo(6))
```

### 5. Ejemplo Adicional: Cálculo de Potencia Recursiva

```python
def potencia(a, b):
    # Caso base: Cualquier número elevado a la potencia 0 es 1
    if b == 0:
        return 1
    # Caso recursivo: Multiplicamos la base 'a' por sí misma 'b' veces
    else:
        return a * potencia(a, b - 1)

print("2 elevado a la potencia 3 es:", potencia(2, 3)) 
# Lógica interna: 2 * (2 * (2 * 1)) = 8
```
