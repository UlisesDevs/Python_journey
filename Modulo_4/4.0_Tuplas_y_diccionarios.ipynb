### 1. Conceptos Previos: Mutabilidad y Datos Secuenciales

Antes de entrar de lleno con estos dos nuevos tipos de datos, debemos repasar un par de conceptos fundamentales:

* **Mutabilidad e inmutabilidad:** Se refiere a si un tipo de dato puede ser alterado en memoria durante la ejecución de nuestro programa. Al objeto que puede cambiar su estado o contenido se le llama **mutable**. Al que, una vez creado, no puede cambiar bajo ninguna circunstancia, se le llama **inmutable**.
* **Datos secuenciales:** Hacen referencia a colecciones de elementos almacenados en un orden específico y continuo, lo que permite que puedan ser iterados elemento por elemento (por ejemplo, mediante un ciclo `for`) o accedidos mediante un índice numérico. Las listas y las cadenas de texto (`strings`) son ejemplos de datos secuenciales.

### 2. Tuplas

Una tupla es una estructura de datos secuencial similar a una lista, pero con dos diferencias clave:
1. Utiliza paréntesis `()` en lugar de corchetes `[]` para su declaración.
2. Es **inmutable**. Una vez creada, no puedes agregar, modificar ni eliminar elementos. Solo puede ser leída.

```python
# Para declarar una tupla podemos hacerlo directamente con paréntesis
tupla_1 = ("a", "b", "c", "d")
print("Esta es la tupla 1: ", tupla_1)

# También se puede declarar solo con comas (empaquetado de tuplas)
tupla_2 = "c", "d", "e", "f"
print("Esta es la tupla 2: ", tupla_2)

# Tupla vacía
tupla_vacia = ()

# Es estrictamente necesario agregar una coma después del primer elemento 
# al crear una tupla de un solo elemento para que el intérprete la reconozca como tal.
tupla_de_un_solo_elemento = (1,)
print("Esta es la tupla de un solo elemento: ", tupla_de_un_solo_elemento)
```

### 3. Acceso a Tuplas

Ya que las tuplas son datos secuenciales, podemos acceder a sus elementos mediante índices (positivos y negativos), realizar *slicing* (particionado) y recorrerlas directamente con un ciclo `for`.

```python
# Acceso por medio de índices similar a las listas
my_tuple = (1, 10, 100, 1000)

print(my_tuple[0])
print(my_tuple[-1])
print(my_tuple[1:])
print(my_tuple[:-2])

# Ya que podemos acceder a la tupla mediante el ciclo for, confirmamos que es secuencial
for elem in my_tuple:
    print(elem)
```

### 4. Diccionarios: Claves y Valores

El diccionario es una estructura de datos **mutable** que almacena información en pares de clave-valor (`key: value`). La clave se encuentra a la izquierda de los dos puntos (`:`) y el valor a la derecha.

> **Anotación del Profesor:**
> En tus apuntes originales indicaste que el diccionario es una "estructura de datos secuencial". Esto es **conceptualmente incorrecto**. Los diccionarios son estructuras de **Mapeo** (implementadas bajo el capó como Tablas Hash). No se accede a ellos por un índice numérico secuencial (`diccionario[0]`), sino buscando directamente su clave. 
> 
> Como futuro desarrollador, debes dominar los diccionarios, ya que en el desarrollo Back End con frameworks como Django o FastAPI, los diccionarios de Python se traducen directamente a la estructura universal de respuestas web: el formato JSON.

```python
# Los nombres son las claves y los números los valores
numeros_de_telefono = {"Ulises": 5566133043, "Julio": 5574652371} 
print("Este es el diccionario de numeros telefonicos: ", numeros_de_telefono)

# También podemos declarar diccionarios solo con cadenas de texto
alumnos = {"alumno_1": "Ulises", "alumno_2": "Julio"}
print("Este es el diccionario de alumnos: ", alumnos)

# Si se quiere declarar un diccionario vacío, solo se escriben las llaves sin contenido
diccionario_vacio = {}
```

### 5. Reglas de los Diccionarios y Acceso a Datos

* **Claves Únicas:** Cada clave debe ser única. No es posible tener una clave duplicada.
* **Claves Inmutables (Hashables):** Una clave puede ser un dato de cualquier tipo inmutable (número entero, flotante, string o incluso una tupla), pero **nunca** un tipo de dato mutable como una lista o un diccionario.
* **Estructura de Pares:** Una lista contiene un conjunto de valores numerados; un diccionario almacena relaciones.
* **Longitud:** La función `len()` regresa la cantidad de pares (clave-valor) existentes.
* **Unidireccionalidad:** Es una herramienta de un solo sentido. Buscas la clave para obtener el valor, no al revés.

```python
# Accediendo al valor de una clave. No se puede utilizar una clave que no exista (lanzará KeyError)
print("El numero de Ulises es: ", numeros_de_telefono["Ulises"])

# Utilizando la sintaxis anterior podemos iterar sobre el diccionario
for alumno in alumnos:
    print(f"El nombre del {alumno} es: ", alumnos[alumno])
```

### 6. Métodos y Funciones de los Diccionarios

Para interactuar con la información estructurada, Python provee métodos específicos:
* `.keys()`: Retorna una vista con todas las claves del diccionario.
* `.values()`: Retorna una vista con todos los valores.
* `.items()`: Retorna una vista compuesta por tuplas, donde cada tupla contiene el par `(clave, valor)`.
* `sorted()`: Función global que permite ordenar los elementos devueltos.

```python
print("Estas son las claves del diccionario de alumnos: ", alumnos.keys())

# .items() retorna tuplas de cada clave y valor
print("Este es el ejemplo de items(): ", alumnos.items())

print("Estos son los valores del diccionario de alumnos: ", alumnos.values())

# sorted() devuelve una lista ordenada
print("Lista ordenada de los valores del diccionario: ", sorted(alumnos.values()))
```

### 7. Modificar, Agregar y Borrar Valores

Ya que los diccionarios son mutables, es sencillo modificar sus valores u operar sobre ellos con distintos métodos integrados.

> **Anotación del Profesor:**
> Borrar con `del diccionario["clave"]` es funcional, pero si la clave no existe, tu programa colapsará. En bases de datos y desarrollo web, es mucho más seguro usar el método `.pop("clave", valor_por_defecto)`.

```python
# Agregar un nuevo valor al diccionario
alumnos["alumno_3"] = "Juan"
print("Diccionario con el nuevo alumno agregado: ", alumnos)

# Modificar un valor ya existente
alumnos["alumno_3"] = "Jose"
print("Diccionario con el alumno modificado: ", alumnos)

# También podemos agregar nuevos valores con el método .update()
alumnos.update({"alumno_4": "Ricardo"})
print("Nuevo alumno agregado mediante update(): ", alumnos)

# O modificar uno ya existente con el mismo método
alumnos.update({"alumno_1": "Mictlan"})
print("Alumno 1 modificado mediante update(): ", alumnos)

# Para borrar un valor utilizamos la palabra clave del
del alumnos["alumno_1"]
print("Alumno 1 borrado mediante la palabra clave del: ", alumnos)

# Para eliminar el último valor insertado utilizamos el método .popitem():
alumnos.popitem()
print("Diccionario con el ultimo valor borrado mediante .popitem():", alumnos)
```

### 8. Integración de Diccionarios y Tuplas

El siguiente programa es un ejemplo de cómo los diccionarios y las tuplas pueden trabajar juntos.
*Nota: Este programa aún no cuenta con manejo de excepciones para los tipos de datos de entrada.*

```python
school_class = {}

while True:
    name = input("Ingresa el nombre del estudiante (o presiona Enter para salir): ")
    if name == '':
        break
    
    score = int(input("Ingresa la calificación del estudiante (0-10): "))
    if score not in range(0, 11):
        break

    # Si el alumno ya existe, concatenamos la nueva calificación a su tupla
    if name in school_class:
        school_class[name] += (score,)
    # Si no existe, creamos la clave y le asignamos una tupla con su primera calificación
    else:
        school_class[name] = (score,)
        
for name in sorted(school_class.keys()):
    adding = 0
    counter = 0
    for score in school_class[name]:
        adding += score
        counter += 1
    
    print(name, ":", adding / counter)
```

## Corrección de Bloques de Código de Recursión
Si deseas limpiar tu sección anterior de recursión, reemplaza tus celdas de código por las siguientes:

```python
#Factorial
def factorial(n):
    # Caso base: Si n es 0 o 1, el factorial es 1
    if n == 0 or n == 1:
        return 1
    # Caso recursivo: n multiplicado por el factorial de n-1
    else:
        return n * factorial(n - 1)

print("El factorial de 5 es:", factorial(5))
```

```python
#Fibonacci Recursivo
def fibonacci(n):
    # Caso base: F(0) = 0, F(1) = 1
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Caso recursivo: Suma de los dos números anteriores en la serie
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print("La posición 6 de la serie Fibonacci es:", fibonacci(6))
```

```python
#Fibonacci Optimizado (Memoization)
def fibonacci_memo(n, memo=None):
    # Si memo no existe en la llamada inicial, lo creamos de forma local (Previene fugas de datos)
    if memo is None:
        memo = {}
        
    # Casos base
    if n == 0:
        return 0
    elif n == 1:
        return 1
        
    # Si el resultado ya fue calculado previamente y está en memoria, lo retornamos directo
    if n in memo:
        return memo[n]
        
    # Caso recursivo: calculamos y almacenamos en la caché local al mismo tiempo
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    
    return memo[n]

print("Fibonacci optimizado (posición 6):", fibonacci_memo(6))
```

```python
#Potencia
def potencia(a, b):
    # Caso base: Cualquier número elevado a la potencia 0 es 1
    if b == 0:
        return 1
    # Caso recursivo: Multiplicamos la base 'a' por sí misma 'b' veces
    else:
        return a * potencia(a, b - 1)

print("2 elevado a la potencia 3 es:", potencia(2, 3))
```
