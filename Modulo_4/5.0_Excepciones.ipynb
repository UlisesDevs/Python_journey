### Módulo: Manejo de Errores y Testing
#### Unidad 1: El Problema de las Excepciones

El manejo de errores hace referencia al comportamiento que puede tener nuestro programa al introducir datos que afecten la ejecución del mismo. Su propósito es atrapar estas anomalías para evitar que el programa termine abruptamente (lo que en un entorno Back End se traduciría en la caída del servidor o un error 500 para el usuario) y protegerlo ante entradas no válidas o bugs.

Por ejemplo, si tenemos un programa que calcula el recíproco de un número mediante una división y el usuario ingresa un `0`, matemáticamente es imposible realizar la operación. Si no manejamos esto, el programa colapsará.

```python
# EJEMPLO SIN MANEJO DE ERRORES
# Si ingresas un 0, una letra o presionas Enter vacío, el programa colapsará.
valor = int(input("Ingresa un número natural para calcular su recíproco: "))
print("El recíproco de", valor, "es", 1 / valor)
```

#### Unidad 2: Bloques `try - except`

Si bien podríamos hacer comprobaciones previas usando condicionales y funciones como `type()`, Python tiene una filosofía de diseño llamada **EAFP** (*"Easier to Ask for Forgiveness than Permission"* - Es más fácil pedir perdón que permiso). 

En lugar de verificar todo antes de actuar, intentamos ejecutar el código y, si ocurre un error, lo atrapamos y lo manejamos con las palabras reservadas `try` y `except`.



* `try`: En este bloque se coloca el código que creemos que es propenso a errores.
* `except`: En este bloque se maneja el error. El código aquí solo se ejecuta si el bloque `try` falla.

```python
# PRIMER MANEJO DE ERRORES (Excepción General)
try:
    valor = int(input("Ingresa un número natural para calcular su recíproco: "))
    print("El recíproco de", valor, "es", 1 / valor) 
except:
    # Este bloque atrapa cualquier error sin distinción
    print("Ocurrió un error. Asegúrate de ingresar un número válido distinto de cero.")
```

#### Unidad 3: Manejo de Excepciones Específicas

Un `except` genérico es una mala práctica profesional porque oculta la verdadera causa del error. Lo correcto es identificar errores concretos y arrojar un texto/acción para cada uno. Podemos utilizar múltiples bloques `except` en una misma estructura.

```python
# MANEJO DE ERRORES ESPECÍFICOS
try:
    valor_str = input("Ingresa un número natural para calcular su recíproco: ")
    valor = int(valor_str)
    print("El recíproco de", valor, "es", 1 / valor) 
except ValueError:
    # Se ejecuta si int() falla al intentar convertir una letra o símbolo
    print(f"'{valor_str}' no es un número válido.")
except ZeroDivisionError:
    # Se ejecuta si el usuario ingresa 0
    print("Matemáticamente, no se puede dividir entre 0.")
```

#### Unidad 4: Capturando el Error y Levantando Excepciones (`raise`)

¿Qué pasa si el usuario presiona Enter sin escribir nada? Esto arroja un `ValueError`, pero sería mejor darle visibilidad exacta de lo que hizo mal. Para esto podemos asignar el error a una variable usando `as e` y también podemos forzar (levantar) nuestros propios errores usando la palabra reservada `raise`.

```python
# ALIAS DE ERRORES Y EXCEPCIONES MANUALES
try:
    valor_str = input("Ingresa un número natural: ")
    
    # Verificamos si la entrada está vacía y levantamos un error manual
    if valor_str == '':  
        raise ValueError("Entrada vacía, se esperaba un número.")
    
    valor = int(valor_str)
    print("El recíproco de", valor, "es", 1 / valor)
    
except ValueError as e:
    # 'e' contiene el mensaje exacto del error. 
    # Comprobamos si es nuestro error manual o un error de conversión de Python.
    if str(e) == "Entrada vacía, se esperaba un número.":
        print(e)  
    else:
        print(f"Debes ingresar un número válido, no puedo procesar '{valor_str}'.")
        
except ZeroDivisionError:
    print("La división entre cero no está permitida en nuestro Universo.")
```

#### Unidad 5: Creando un Ciclo Robusto

En aplicaciones de consola, podemos utilizar un bucle `while` para atrapar al usuario hasta que ingrese los datos correctamente.

```python
# BUCLE DE VALIDACIÓN ROBUSTO
while True:
    try:
        valor_str = input("Ingresa un número natural (o escribe 'salir'): ")
        
        if valor_str.lower() == 'salir':
            print("Saliendo del programa...")
            break
            
        if valor_str == '':  
            raise ValueError("Entrada vacía, se esperaba un número.")
        
        valor = int(valor_str)
        print("El recíproco de", valor, "es", 1 / valor)
        
        # Si llegamos a esta línea sin errores, rompemos el ciclo con break
        break 
        
    except ValueError as e:
        if str(e) == "Entrada vacía, se esperaba un número.":
            print(e)
        else:
            print(f"Error: No puedo procesar '{valor_str}'. Intenta de nuevo.")
            
    except ZeroDivisionError:
        print("Error: No puedes ingresar 0. Intenta de nuevo.")
```

#### Unidad 6: Diccionario de Excepciones Comunes

| Excepción | Descripción y Contexto |
| :--- | :--- |
| **`ZeroDivisionError`** | Aparece cuando intentas forzar una división donde el divisor es cero. Puede ser detonada por los operadores `/`, `//` o `%`. |
| **`ValueError`** | Se genera cuando una función (como `int()` o `float()`) recibe un argumento de un tipo adecuado, pero su valor es inaceptable (ej. intentar convertir la letra "A" a entero). |
| **`TypeError`** | Ocurre al intentar aplicar una operación a un dato de tipo incompatible. Ejemplo: usar un número flotante como índice de una lista (`lista[0.5]`) o intentar sumar un entero con un string (`5 + "hola"`). |
| **`AttributeError`** | Llega cuando intentas activar un método o atributo que no existe en el elemento. Ejemplo: intentar usar `.append()` en una tupla o `.depend()` en una lista. |

### Módulo: Pruebas y Depuración (Testing & Debugging)

Probar el código es esencial. Las pruebas no buscan demostrar que el código está libre de errores, sino forzarlo para descubrir dónde falla. Los desarrolladores tienden a ser subjetivos con su propio trabajo, por lo que una mentalidad destructiva al momento de probar es vital.

* **Depuración (Debugging):** Es el proceso de aislar y eliminar errores lógicos del código.
* **Depuradores Interactivos:** Herramientas (incluidas en editores como VSCode) que permiten ejecutar el programa línea por línea, inspeccionar el estado de las variables en memoria RAM y detener la ejecución mediante *breakpoints*.
* **Depuración por Impresión:** Técnica básica donde se insertan declaraciones `print()` para observar el flujo de los datos. 

> **Anotación del Profesor (Metodología DevOps y Back End):**
> Aunque la "depuración por impresión" (`print()`) es útil para scripts rápidos, en el desarrollo Back End está estrictamente prohibida en código de producción. En su lugar, debes utilizar el módulo **`logging`** nativo de Python. Los *logs* te permiten guardar registros de errores en archivos de texto, clasificados por severidad (`INFO`, `WARNING`, `ERROR`, `CRITICAL`), los cuales son esenciales para monitorear contenedores en Docker sin tener que ver la consola en tiempo real.


#### Consejos de Depuración Práctica:
1. **Rubber Duck Debugging:** Explica tu código línea por línea en voz alta a un objeto inanimado (como un patito de goma). Verbalizar la lógica a menudo revela fallas estructurales.
2. **Aislamiento:** Comenta partes del código o separa el proyecto por módulos para ejecutar solo el área sospechosa.
3. **Control de Versiones:** Apóyate en Git para revisar los cambios recientes (`git diff`). El error probablemente está en las líneas modificadas en el último *commit*.
4. **Descanso:** Tómate un tiempo para despejar la mente. La fatiga nubla el análisis lógico.

#### Pruebas Unitarias (Unit Testing):

Las pruebas unitarias son *scripts* de código escritos específicamente para probar otras funciones de tu código principal. Python incluye los módulos `unittest` y `pytest` para facilitar este enfoque. 

**Enfoque DevOps:** Agregar pruebas unitarias automatizadas es el núcleo de la Integración Continua (CI). En un flujo moderno, cuando escribes código y haces un *push* a GitHub, una herramienta (como GitHub Actions o Jenkins) levantará un contenedor de Docker temporal, instalará tu entorno, ejecutará todas tus pruebas unitarias y, si alguna falla, bloqueará la subida del código a producción, garantizando que el software entregado sea robusto y confiable.
