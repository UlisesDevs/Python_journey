### 1. Alcances o Scopes en Python

En Python, el **scope** (alcance) se refiere al contexto en el que las variables y nombres son visibles y accesibles. Determina qué variables se pueden usar en diferentes partes de tu código. 

Python maneja el scope de manera jerárquica y estática, lo que significa que cada variable tiene un alcance determinado en el momento en que el código es compilado, y no cambia durante la ejecución.


#### Tipos de Scope en Python (Regla LEGB)

Python utiliza una regla conocida como **LEGB** para determinar el orden en el que busca y resuelve el alcance de las variables:

1. **Local (L):** El scope local es el que pertenece a una función. Las variables definidas dentro de una función solo son accesibles desde esa función.
2. **Enclosing (E):** Si una función está anidada dentro de otra, el scope *enclosing* (o no local) es el que pertenece a la función externa que envuelve a la función interna.
3. **Global (G):** Las variables globales son las que están definidas en el nivel principal del módulo (es decir, fuera de todas las funciones y clases).
4. **Built-in (B):** Estas son las variables y funciones que vienen predefinidas e integradas en Python, como `print()`, `len()`, `sum()`, etc.

```python
# 1. SCOPE LOCAL
def funcion_local():
    x = 10  # 'x' es una variable local, nace y muere dentro de esta función
    print("Local:", x)

funcion_local()


# 2. SCOPE ENCLOSING (Anidado)
def funcion_externa():
    x = 20  # 'x' pertenece al scope de funcion_externa
    
    def funcion_interna():
        # 'x' no es local de funcion_interna, así que Python busca una capa arriba (Enclosing)
        print("Enclosing:", x)  
        
    funcion_interna()

funcion_externa()


# 3. SCOPE GLOBAL
x_global = 30  # 'x_global' es una variable global

def funcion_global():
    # Python busca localmente, luego enclosing, y al no encontrarla, la toma del Global
    print("Global (Solo lectura):", x_global)  

funcion_global()
```

### 2. Modificando Variables Globales y Enclosing

Por defecto, dentro de una función solo puedes **leer** una variable global, no modificarla. Si intentas reasignarle un valor, Python creará una variable local nueva con el mismo nombre (Shadowing).

* Para modificar una variable global desde dentro de una función, debes usar la palabra clave `global`.
* Si estás trabajando con funciones anidadas y quieres modificar una variable en el *enclosing scope* (el scope de la función externa), debes usar la palabra clave `nonlocal`.

> **Anotación del Profesor:**
> **Evita el uso de `global` en producción.** En el desarrollo Back End, modificar variables globales desde dentro de funciones se considera una mala práctica arquitectónica porque genera "Efectos Secundarios" (Side Effects). Esto hace que el código sea difícil de testear y propenso a errores (bugs) inesperados. 
> 
> Una buena función debe ser una "Función Pura": recibe datos por los parámetros, los procesa y los devuelve con un `return`, sin modificar nada del exterior.

```python
# MODIFICANDO VARIABLES GLOBALES
x_modificable = 100  # Variable global

def modificar_global():
    global x_modificable  # Le indicamos a Python explícitamente que usaremos la variable global
    x_modificable = 200   # Ahora sí se modifica la original

modificar_global()
print("Global modificada:", x_modificable)


# MODIFICANDO VARIABLES ENCLOSING (NONLOCAL)
def funcion_externa_mod():
    y = 50
    
    def funcion_interna_mod():
        nonlocal y  # Apunta a la variable 'y' del scope de la función externa
        y = 75
        
    funcion_interna_mod()
    print("Enclosing modificada con nonlocal:", y)

funcion_externa_mod()
```
