### 1. ¿Qué es una Función?

Una función es un bloque de código que puede ser invocado en cualquier parte de nuestro programa las veces que lo necesitemos. Nos sirve para evitar reescribir código a menudo y simplifica la comprensión de nuestro programa.

Existen distintas razones para trabajar con funciones:
* **Reutilización (Principio DRY - Don't Repeat Yourself):** Si un fragmento de código necesita utilizarse muchas veces, no es necesario volver a escribirlo. Con definir una vez nuestra función e invocarla será suficiente.
* **Descomposición:** Si tu código se vuelve tan extenso que leerlo y comprenderlo se hace complicado, las funciones permiten descomponer un gran problema en problemas más pequeños. Esta descomposición continúa hasta que tienes un conjunto de funciones cortas, fáciles de comprender y probar.
* **Delegación:** Permite dividir el trabajo entre distintos desarrolladores, ya que cada uno puede escribir un conjunto bien definido de funciones que, al integrarse, resuelven el problema en común.

**¿De dónde provienen las funciones?**
1. **Integradas (Built-in):** Provienen de Python mismo (como `print()`, `len()`) y están a nuestra disposición en cualquier momento.
2. **De Módulos:** Estas funciones necesitan ser importadas primero para poder invocarlas (ej. `import math`).
3. **Definidas por el usuario:** Escritas por nosotros mismos. Definiremos la función y la invocaremos cuando la necesitemos.

### 2. Sintaxis y Definición Básica

* Una función se define con la palabra reservada `def`.
* Después un espacio y el nombre de la función (sigue las mismas reglas que las variables).
* A continuación, se ponen paréntesis `()`, dentro de los cuales escribiremos los parámetros en caso de necesitarlos.
* Se termina la línea con dos puntos `:`.
* El cuerpo de nuestra función irá indentado en la línea siguiente. La función termina donde termine nuestra indentación.

> **Anotación del Profesor:**
> Definir una función no ejecuta su código. Para que el código corra, debes **invocarla** (llamarla) escribiendo su nombre seguido de paréntesis. No puedes invocar una función antes de que el intérprete de Python haya leído su definición.

```python
# DEFINICIÓN
def saludo():
    print("¡Hola, Mundo!")

# INVOCACIÓN
saludo()
```

### 3. Funciones con Parámetros y Argumentos

Para definir una función que recibe datos, los parámetros se declaran dentro de los paréntesis al momento de definirla.

**Nota técnica:** 
* **Parámetros:** Son las variables declaradas en la definición de la función. Son de ámbito local (solo existen dentro de la función).
* **Argumentos:** Son los valores reales que pasamos a la función al momento de invocarla.

```python
def saludo_2(nombre):
    # 'nombre' es el parámetro
    print("Hola,", nombre)

# "Ulises" es el argumento
saludo_2("Ulises")
```

### 4. Argumentos Posicionales, de Palabra Clave y Predefinidos

Una función puede contener tantos parámetros como necesitemos (aunque por buenas prácticas de Clean Code, si tienes más de 3 o 4 parámetros, deberías considerar pasar un objeto o diccionario).

* **Posicionales:** Los argumentos se asignan en el orden exacto en que se declararon los parámetros.
* **Palabra Clave (Keyword Arguments):** Se asignan mencionando explícitamente el nombre del parámetro, sin importar la posición.
* **Predefinidos (Default):** Se asignan valores por defecto en la definición. Si el usuario no envía un argumento, la función usa el predefinido.

```python
def hobby(nombre, pasatiempo):
    print("A", nombre, "le gusta", pasatiempo)

# 1. Argumentos Posicionales
hobby("Ulises", "Programar")

# 2. Argumentos de Palabra Clave (No importa el orden)
hobby(pasatiempo="cantar", nombre="Ulises")

# 3. Combinación (Los posicionales siempre deben ir primero)
def concat(a, b, c):
    print(a, "+", b, "+", c, "=", a + b + c)

concat(1, c=3, b=4)

# 4. Parámetros Predefinidos
def pelicula(nombre="Ulises", titulo="Pulp Fiction"):
    print("La película favorita de", nombre, "es", titulo)

print("\n--- Pruebas con Predefinidos ---")
pelicula() # Usa ambos predefinidos
pelicula("Juan", "Spiderman") # Sobrescribe ambos
pelicula(titulo="Harry Potter") # Sobrescribe solo el título
```

### 5. La Sentencia `return` (Retornando Resultados)

Si queremos que una función procese datos y nos devuelva un valor para usarlo en otra parte del programa, utilizamos la palabra reservada `return`. Al ejecutarse un `return`, la función termina inmediatamente.

> **Anotación del Profesor:**
> **¡CUIDADO!** En tus apuntes originales escribiste: `resultado = print(numero_1 + numero_2)`. Esto es un error grave. La función `print()` sirve únicamente para mostrar texto en la consola de tu monitor; **no procesa ni almacena valores**. La función `print()` siempre devuelve el tipo de dato `None`. Si haces `return print()`, tu función devolverá la "nada" (None) y tu programa fallará si intentas usar ese resultado para otra operación matemática.
>
> La forma correcta es calcular el valor, guardarlo en una variable (o retornarlo directamente) y usar el `print()` por fuera si necesitas verlo. Observa las correcciones en la siguiente celda.

```python
# FORMA CORRECTA DE RETORNAR UN VALOR
def suma(numero_1, numero_2):
    resultado = numero_1 + numero_2
    return resultado

valor_suma = suma(23, 45)
print("El resultado devuelto de la suma es:", valor_suma)

# FORMA CORRECTA DE RETORNAR MÚLTIPLES VALORES (Retorna una Tupla)
def operacion(numero_1, numero_2):
    res_suma = numero_1 + numero_2
    res_resta = numero_1 - numero_2
    return res_suma, res_resta 

# Desempaquetado de los valores devueltos
total_suma, total_resta = operacion(34, 89)
print("\nLa suma devuelta es:", total_suma)
print("La resta devuelta es:", total_resta)

# RETURN DENTRO DE CONDICIONALES
def evaluar_numero(x):
    if x > 0:
        return "Positivo"
    elif x < 0:
        return "Negativo"
    else:
        return "Cero"

print("\nEvaluaciones:", evaluar_numero(5), ",", evaluar_numero(-2), ",", evaluar_numero(0))
```

### 6. Pasando Estructuras de Datos como Argumentos

En una función podemos pasar listas, diccionarios o cualquier estructura como argumento.


```python
def sumar_elementos_lista(lista_numeros):
    acumulador = 0
    for elemento in lista_numeros:
        acumulador += elemento
    return acumulador

mi_arreglo = [5, 4, 3]
resultado_arreglo = sumar_elementos_lista(mi_arreglo)

print("La suma de los valores de la lista es:", resultado_arreglo)

# Recordatorio de Back End: En producción usarías directamente sum(mi_arreglo)
```
